```python
# Email Phishing Classifier - Test Notebook

import tensorflow as tf
import numpy as np
import os

possible_model_paths = [
    '../Model/email_classifier_final.keras',
    './Model/email_classifier_final.keras',
    'Model/email_classifier_final.keras',
    'email_classifier_final.keras'
]

model_path = None
for path in possible_model_paths:
    if os.path.exists(path):
        model_path = path
        break

if model_path is None:
    raise FileNotFoundError(
        "Could not find email_classifier_final.keras. "
        "Please ensure the model file is in the correct location."
    )

print(f"Loading model from: {model_path}")
model = tf.keras.models.load_model(model_path)
print("✅ Model loaded successfully!\n")



def predict_email(text):
    """
    Predict whether an email is legitimate or malicious.
    
    Args:
        text (str or list): Single email text or list of email texts
    
    Returns:
        tuple: (predictions, probabilities)
    """
    # Handle single string input
    if isinstance(text, str):
        text = [text]
    
    # Convert to tensor and get predictions
    text_tensor = tf.constant(text)
    probabilities = model(text_tensor, training=False).numpy().flatten()
    
    # Convert probabilities to labels
    predictions = ["🚨 MALICIOUS" if p > 0.5 else "✅ LEGITIMATE" for p in probabilities]
    
    return predictions, probabilities


def test_single_email(email_text):
    """
    Test a single email and display detailed results.
    
    Args:
        email_text (str): The email text to classify
    """
    prediction, probability = predict_email(email_text)
    prob_value = probability[0]
    
    print("\n" + "="*80)
    print("PREDICTION RESULT")
    print("="*80)
    print(f"\n📧 Email Text:")
    print(f"   {email_text}")
    print(f"\n🔍 Prediction: {prediction[0]}")
    print(f"📊 Malicious Probability: {prob_value:.2%}")
    print(f"📊 Legitimate Probability: {(1-prob_value):.2%}")
    
    # Visual confidence bar
    bar_length = 50
    mal_bar = int(prob_value * bar_length)
    legit_bar = bar_length - mal_bar
    
    print(f"\n📈 Confidence Visualization:")
    print(f"   Malicious:  [{'█' * mal_bar}{'░' * legit_bar}] {prob_value:.1%}")
    print(f"   Legit: [{'█' * legit_bar}{'░' * mal_bar}] {(1-prob_value):.1%}")
    print("="*80 + "\n")



def interactive_predict():
    """
    Interactive prediction function for Jupyter notebooks.
    Prompts user for input and displays prediction.
    """
    print("Enter an email to classify (or 'quit' to exit):")
    
    while True:
        user_input = input("\n📧 Email text: ").strip()
        
        if user_input.lower() in ['quit', 'exit', 'q']:
            print("Exiting...")
            break
        
        if not user_input:
            print("Please enter some text!")
            continue
        
        test_single_email(user_input)
        print("\nEnter another email (or 'quit' to exit):")


interactive_predict()
```
